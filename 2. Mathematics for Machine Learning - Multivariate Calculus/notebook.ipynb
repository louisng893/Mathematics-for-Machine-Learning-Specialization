#### Definition of a derivative

gradient (or slope) is "rise over run" $ = \frac{rise}{run} = \frac{y_1 - y_0}{x_1 - x_0}$

$$
\text{gradient at x} = \lim_{\Delta x \to 0}\frac{f(x+\Delta x) - f(x)}{\Delta x}
$$
or in short:
$$ 
\boxed{
    \frac{df}{dx} = f'(x) = \lim_{\Delta x \to 0}\frac{f(x+\Delta x) - f(x)}{\Delta x}
}
$$

*Example:* $f(x) = 3x + 2$
$$
\begin{align*}
f'(x) &= \lim_{\Delta x \to 0}\frac{3(x+\Delta x)+2 -(3x+2)}{\Delta x} \\
&= \lim_{\Delta x \to 0}\frac{3x+3\Delta x+2 -3x+2}{\Delta x} = \lim_{\Delta x \to 0}\frac{3\Delta x}{\Delta x} \\
&= \lim_{\Delta x \to 0}3 = 3
\end{align*}
$$

#### Defferentiation example & cases
- $f(x) = \frac{1}{x}$
$$
\begin{align*}
f'(x) &= \lim_{\Delta x \to 0}\frac{\frac{1}{x + \Delta x} - \frac{1}{x}}{\Delta x} \\
&= \lim_{\Delta x \to 0}\frac{\frac{x - x - \Delta x}{x(x + \Delta x)}} {\Delta x} \\
& = \lim_{\Delta x \to 0}\frac{-1}{x^2 + x\Delta x}\\
& = \frac{-1}{x^2}\\
\end{align*}
$$
- $f(x) = f'(x)$ at $f(x) = f'(x) = f''(x) = f^{(3)}(x) e^x$

### Time saving rules
- Sum rule:
$$
\frac{d}{dx}(f(x) + g(x)) = \frac{df(x)}{dx} + \frac{dg(x)}{dx}
$$
- Power rule: If $f(x) = ax^b$ then $f(x) = abx^{b-1}$
- Product rule:
$$
\begin{align*}
\frac{d}{dx}(f(x)g(x)) &= \frac{df(x)}{dx}g(x) + f(x)\frac{dg(x)}{dx} \\
\iff [f(x)g(x)]' &= f'(x)g(x) + f(x)g'(x)
\end{align*}
$$
- Chain rule: If $f = f(g)$ and $g = g(x)$, then
$$
\frac{df}{dx} = \frac{df}{dg}\frac{dg}{dx}
$$

*Example:* $f(x) = \frac{sin(2x^5+3x)}{e^{7x}} = (sin(2x^5 + 3x))e^{-7x}$
$$
\begin{align*}
\frac{df}{dx}  &= \frac{dg}{dx}h + g\frac{dh}{dx} \quad\text{(product rule)}\\
&= (10x^4 + 3)cos(2x^5+3x)e^{-7x} + sin(2x^5 + 3x)(-7e^{-7x}) \quad\text{(chain rule)} \\
&= e^{-7x}[(10x^4 +3)cos(2x^5 + 3x) - 7sin(2x^5 + 3x)] \\[10pt]
&= \frac{(10x^4 + 3)cos(2x^5 + 3x)}{e^{7x}} - 7f(x)
\end{align*}

### Jacobians - vectors of derivatives

#### The Jacobian
We consider a single function with many variables $f(x_1, x_2, x_3,...)$, then the Jacobian is simply a vector where each entry is the partial derivative of $f$ (with row vectors rather than column vectors).
$$
J = \begin{bmatrix}\frac{\partial f}{\partial x_1}&\frac{\partial f}{\partial x_2}&\frac{\partial f}{\partial x_3}&\cdots\end{bmatrix}
$$
*Example:* $f(x, y, z) = x^2y + 3z$
$$
\left. 
\begin{align*}
\frac{\partial f}{\partial x} &= 2xy \\
\frac{\partial f}{\partial y} &= x^2 \\
\frac{\partial f}{\partial z} &= 3
\end{align*}
\right\} \qquad J = \begin{bmatrix}2xy&x^2&3\end{bmatrix}
$$
For a specific given $x$, $y$, and $z$ will return a vector pointing in the direction of steepest slope of this function. For example, at $J(0, 0, 0) = \begin{bmatrix}0&0&3\end{bmatrix}$. 
> The larger the magnitude of the Jacobian vectors, the steeper the slope.

#### Jacobian Applied
$$
\boxed{
    J = \begin{bmatrix} \frac{\partial u}{\partial x}&\frac{\partial u}{\partial y}\\\frac{\partial v}{\partial x}&\frac{\partial v}{\partial y}\end{bmatrix}
    }
$$
For the pair of equations: 
- $u(x, y) = x - 2y$
- $v(x, y) = 3y - 2x$

we can calculate the Jacobian matrix as: $J = \begin{bmatrix}1&-2\\-2&3\end{bmatrix}$, there are no longer any variables in the Jacobian matrix, it means both $u$ and $v$ are linear functions of $x$ and $y$, so the gradient must be constant everywhere. Also, this matrix is just the linear transformation from $x\circ y$ space to $u\circ v$ space.

If we consider a vector $\begin{pmatrix}2\\3\end{pmatrix}$, we get the following $\begin{bmatrix}1&-2\\-2&3\end{bmatrix}\begin{pmatrix}2\\3\end{pmatrix} = \begin{pmatrix}-4\\5\end{pmatrix}$.

Sometime functions will be highly nonlinear, much more complicated. However, they may still be smooth, which means that if we zoom in close enough, we can consider each little region of space to be approximately linear. Therefore, by adding up all the contributions from the Jacobian determinants at each point in space, we can still calculate the change in the size of a region after transformation.

***Example:*** transform between Cartesian and Polar coodinate systems



If we have a vector expressed in terms of a radius $r$, and the angle up $\theta$ from the x-axis. But we would like them expressed in terms of $x$ and $y$ instead.
- $x(r, \theta) = rcos(\theta)$
- $y(r, \theta) = rsin(\theta)$

Then the Jacobian matrix is
$$ 
\begin{align*}
J &= \begin{bmatrix}\frac{\partial x}{\partial r}&\frac{\partial x}{\partial\theta}\\\frac{\partial y}{\partial r}&\frac{\partial y}{\partial\theta}\end{bmatrix}\\
& =\begin{bmatrix}cos(\theta)&-rsin(\theta)\\ sin(\theta)&rcos(\theta)\end{bmatrix} \\[10pt]
\implies |J| &= r(cos^2(\theta) + sin^2(\theta)) \\
&= r
\end{align*}
$$


### The sandpit game

#### The sandpit game
We consider a function $f(x, y) = e^{-(x^2+y^2)}$, then we can easily find the Jacobian as discussed $J = \begin{bmatrix}-2xe^{-(x^2 + y^2)}&-2ye^{-(x^2 + y^2)}\end{bmatrix} = 0$ (it also means $f'(x, y) = 0$), solving this equation we get $x = 0$ and $y = 0$.

Therefore, with this function or more other complicated functions, we have, at the critical points - including minima, maxima, or saddle point,the Jacobian (gradient) equals to zero ($J=0$). The direction of the vector points to the direction of the steepest ascent (uphill), and its length indicates how steep the slope is (at a point). In general, if we stand near **the local minima** (the trough), the gradient vectors around us will **point away from** us. Conversely, near **the local maxima** (the peak), around gradient vectors will **point toward** us.

#### The Hessian
For the Jacobian, we collected together all *the first order* derivatives of a function into a vector. We are going to collect all of *the second order* derivatives together into a matrix, which for a function of n variables - **The Hessian**.

$$
\boxed{
    \mathbf{H} = \begin{bmatrix}\frac{\partial^2 f}{\partial x_{1}^{2}}&\frac{\partial^2 f}{\partial x_{1}x_{2}}&\cdots&\frac{\partial^2 f}{\partial x_{1}x_{n}}\\[5pt]\frac{\partial^2 f}{\partial x_{2}x_{1}}&\frac{\partial^2 f}{\partial x_{2}^{2}}&\cdots&\frac{\partial^2 f}{\partial x_{2}x_{n}}\\[5pt]\vdots&\vdots&\ddots&\vdots\\[5pt]\frac{\partial^2 f}{\partial x_{n}x_{1}}&\frac{\partial^2 f}{\partial x_{n}^{2}}&\cdots&\frac{\partial^2 f}{\partial x_{n}^{2}}\end{bmatrix}
}
$$
We can write it using an abbreviated notation style
$$
\iff 
\mathbf{H} = \begin{bmatrix}\partial_{x_1x_1}f&\partial_{x_1x_2}f&\cdots&\partial_{x_1x_n}f\\\partial_{x_2x_1}f&\partial_{x_2x_2}f&\cdots&\partial_{x_2x_n}f\\\vdots&\vdots&\ddots&\vdots\\\partial_{x_nx_1}f&\partial_{x_nx_2}f&\cdots&\partial_{x_nx_n}f\end{bmatrix}
$$
The Hessian matrix is an n by n square matrix, where n is the number of variables in our function $f$.

*Example:* $f(x, y, z) = x^2yz$ (find the Jacobian first, then the Hessian)
$$
\begin{align*}
J &=\begin{bmatrix}2xyz&x^2z&x^2y\end{bmatrix} \\[10pt]
H &= \begin{bmatrix}2yz&2xz&2xy\\2xz&0&x^2\\2xy&x^2&0\end{bmatrix}
\end{align*}
$$
Our Hessian matrix is symmetrical across the leading diagonal. This will always be true if the function is continuous, meaning that it has no sudden step changes.

**For example:** We consider two cases:
- Case 1: $f(x, y) = x^2 + y^2$, then the Jacobian is $J = \begin{bmatrix}2x&2y\end{bmatrix}$, and the Hessian is $\mathbf{H} = \begin{bmatrix}2&0\\0&2\end{bmatrix}$. Therefore, the determinant of the Hessian ($|\mathbf{H}| = 4$) is **positive**, we are dealing with **either a maximum or a minimum** (or an optimum point). Moreover, the first term at top left of the Hessian, ($\mathbf{H} = \begin{bmatrix}\mathbf{2}&0\\0&2\end{bmatrix}$), is also **positive** then we have got **a minimum** (when it is negative, we have got a maximum).

    

- Case 2: $f(x, y) = x^2 - y^2$, then the Jacobian is $J = \begin{bmatrix}2x&-2y\end{bmatrix}$, and the Hessian is $\mathbf{H} = \begin{bmatrix}2&0\\0&-2\end{bmatrix}$. Therefore, the determinant of the Hessian ($|\mathbf{H}| = -4$) is **negative**, we are **not** dealing with **a maximum or a minimum**. At this point, the gradient is 0, but with slopes coming down towards it in one direction, but up towards it in the other - **the saddle point**.

    

#### Reality is hard


In 2D space, we can approximate the gradient at x as discussed:
$$
\text{Gradient at x} \approx \frac{f(x+\Delta x)-f(x)}{\Delta x}
$$
However, that is not going to be practical for high-dimensional scenarios.



$$
J = \begin{bmatrix}\frac{f(x+\Delta x,y)-f(x,y)}{\Delta x}&\frac{f(x,y + \Delta y)-f(x, y)}{\Delta y}\end{bmatrix}
$$
If our data is a bit **noisy** (function change to often in small variables). To deal with this case, many different approaches have been developed. But perhaps the simplest is just to calculate the gradient using **a few different step sizes** and take some kind of **average**.

### Chain rule intro.

#### Multivariate chain rule
We consider again the function $f(x, y, z) = sin(x)e^{yz^2}$, and $x=t-1$, $y=t^2$, $z=1/t$. We have the total derivative of $f$ with respect to $t$ by the partial derivative of each variable with the derivative of these variable with variable $t$.
$$
\frac{df}{dt} = \frac{\partial f}{\partial x}\frac{dx}{dt} + \frac{\partial f}{\partial y}\frac{dy}{dt} + \frac{\partial f}{\partial z}\frac{dz}{dt}
$$
This allowed as to calculate the result in a Piecewise manner rather than substituting everything in at the start.
$$
\frac{df}{dt} = cos(t-1)e
$$


In general, we consider the function in which $\mathbf{x}$ is a sequence of variables.
$$
f(x_1, x_2, x_3, ..., x_n) = f(\mathbf{x})
$$
each of the components of our $\mathbf{x}$ vector are themselves functions of some other variable $t$.
$$
\begin{align*}
x_1(t) &= \cdots \\
x_2(t) &= \cdots \\
x_3(t) &= \cdots \\
&\vdots \\
x_n(t) &= \cdots \\
\end{align*}
$$
and what we would like to know is the derivative of $f$ with respect to $t$.
$$
\frac{df}{dt} = ?
$$

In conclusion, we have the following chain rule for multivariate or the multivariate chain rule:
$$
f(\mathbf{x}) = f(x_1, x_2, x_3, ..., x_n)
$$
the partial derivative of $f$ with respect to each components of $\mathbf{x}$:
$$
\frac{\partial f}{\partial \mathbf{x}} = \begin{bmatrix}\frac{\partial f}{\partial x_1}\\[5pt]\frac{\partial f}{\partial x_2}\\[5pt]\frac{\partial f}{\partial x_3}\\[5pt]\vdots\\[5pt]\frac{\partial f}{\partial x_n}\end{bmatrix}
$$
as well as the derivatives of each components of $\mathbf{x}$ with respect to $t$:
$$
\frac{d\mathbf{x}}{dt} = \begin{bmatrix}\frac{dx_1}{dt}\\[5pt]\frac{dx_2}{dt}\\[5pt]\frac{dx_3}{dt}\\[5pt]\vdots\\[5pt]\frac{dx_n}{dt}\end{bmatrix}
$$
we store those derivatives in a pair of n-dimensional vectors. Finally, we are looking to build a multi-variable chain rule expression (the sum of the product of each pair of terms = **the dot product** in **Linear Algebra**):
$$
\begin{align*}
\frac{df}{dt} &= \begin{bmatrix}\frac{\partial f}{\partial x_1}\\[5pt]\frac{\partial f}{\partial x_2}\\[5pt]\frac{\partial f}{\partial x_3}\\[5pt]\vdots\\[5pt]\frac{\partial f}{\partial x_n}\end{bmatrix} \cdot \begin{bmatrix}\frac{dx_1}{dt}\\[5pt]\frac{dx_2}{dt}\\[5pt]\frac{dx_3}{dt}\\[5pt]\vdots\\[5pt]\frac{dx_n}{dt}\end{bmatrix} \\[15pt]
&= \frac{\partial f}{\partial \mathbf{x}} \cdot \frac{dx}{dt} \quad \text{(shortly)}
\end{align*}
$$
We have the generalised form of the multi-variable chain rule as below:
$$
f(x_1, x_2, x_3, ..., x_n) = f(\mathbf{x})
$$
$$
\boxed{
    \frac{df}{dt} = \frac{\partial f}{\partial \mathbf{x}} \cdot \frac{dx}{dt}
}
$$

#### More multivariate chain rule
We have the multivariate chain rule (compact form) as
$$
\begin{align*}
&f(\mathbf{x}(t)) \\
\frac{df}{dt} &= \frac{\partial f}{\partial \mathbf{x}} \cdot \frac{\partial\mathbf{x}}{dt}
\end{align*}
$$
the partial derivative of function $f$ with respect to each component of $\mathbf{x}$ is just the same as a Jacobian vector, but we wrote as a column instead of a raw vector:
$$
\frac{\partial f}{\partial \mathbf{x}} = \begin{bmatrix}\frac{\partial f}{\partial x_1}\\[5pt]\frac{\partial f}{\partial x_2}\\[5pt]\frac{\partial f}{\partial x_3}\\[5pt]\vdots\\[5pt]\frac{\partial f}{\partial x_n}\end{bmatrix}
$$
or from knowledge of the **Linear Algebra**, we can say that $\frac{\partial f}{\partial \mathbf{x}}$ must be the transverse of the Jacobian of $f$:
$$
\frac{\partial f}{\partial \mathbf{x}} = (J_f)^T
$$
and the last thing is that taking the dot product of two columns vectors is the same operation as multiplying a row vector by a column vector. Finally, the Jacobian offer us the most convenient representation of the multivariate chain rule:
$$
\frac{df}{dt} = J_f\frac{d\mathbf{x}}{dt}
$$

**Example:** 
- we consider to prove the chain rule still works for more than two links (**in univiriate functions**)
$$
\begin{align*}
f(x) &= 5x \\
x(u) &= 1 - u \\
u(t) &= t^2 \\[5pt]
\iff f(t) &= 5(1-t^2) \\
&= 5 - 5t^2
\end{align*}
$$
then we can of course directly differentiate this function $f(t)$:
$$
\frac{df}{dt} = -10t
$$
or we can apply a two-step chain rule as following:
$$
\left.
\begin{align*}
f(x) &= 5x \\[5pt]
x(u) &= 1 - u \\[5pt]
u(t) &= t^2 
\end{align*}
\right\}
\quad \frac{df}{dt} = \frac{df}{dx}\frac{dx}{du}\frac{du}{dt} = (5)(-1)(2t) = -10t
$$

- we will consider the chains rule in **multivariate functions**, $f(\mathbf{x}(\mathbf{u}(t)))$, in which
$$
\left.
\begin{align*}
f(\mathbf{x}) &= f(x_1, x_2) \\[5pt]
\mathbf{x}(\mathbf{u}) &= \begin{bmatrix}x_1(u_1, u_2)\\x_2(u_1, u_2)\end{bmatrix} \\[5pt]
\mathbf{u}(t) &= \begin{bmatrix}u_1(t)\\u_2(t)\end{bmatrix}
\end{align*}
\right\}
\quad \frac{df}{dt} = \frac{\partial f}{\partial \mathbf{x}}\frac{\partial \mathbf{x}}{\partial \mathbf{u}}\frac{d\mathbf{u}}{dt}
$$
We have already known differentiating the scale of the valued function $f$, with respect to its input vector $\mathbf{x}$ (multivariate), gives us **the Jacobian** row vector. We have also seen that differentiating a vector valued function $u$ with respect to the scalar variable $t$ (univariate), gives us **a column vector of derivatives**. We need to find the derivative of each of *the two outut variables* with respect to each of *the two input variables*. So we end up with four terms in total as we saw as a matrix (**the Jacobian matrix**):

**Multivariate Chain Rule:**
$$
\iff \boxed{
    \begin{align*}
    \frac{df}{dt} &= \frac{\partial f}{\partial \mathbf{x}}\frac{\partial \mathbf{x}}{\partial \mathbf{u}}\frac{d\mathbf{u}}{dt} = \begin{bmatrix}\frac{\partial f}{\partial x_1}&\frac{\partial f}{\partial x_2}\end{bmatrix} \begin{bmatrix}\frac{\partial x_1}{\partial u_1}&\frac{\partial x_1}{\partial u_2}\\[5pt]\frac{\partial x_2}{\partial u_1}&\frac{\partial x_2}{\partial u_2}\end{bmatrix} \begin{bmatrix}\frac{du_1}{dt}\\[5pt]\frac{du_2}{dt}\end{bmatrix} \\[15pt]
    &= \text{(the Jacobian vector of f)}\times\text{(the Jacobian matrix x)}\times\text{(the derivative vector of u)} \\[10pt]
    (1\times 1) &= (1\times 2)(2\times 2)(2\times 1) \quad \text{(return a scalar)}
    \end{align*}
}
$$



### Neural Networks - Applications of Multivariate Chain Rule

The Neural Networks works that the $\sigma$ gives the neural networks their association to the brain. Neurons in the brain receive information from their neighbours through chemical and electrical stimulation. And when the sum of all these stimulations **goes beyond** a certain threshold amount, the neuron is suddenly **activated** adn **starts stimulating** its neighbours in turn. An example of a function which has this threshold holding property is **hyperbolic tangent function**, $tanh$, which is a nice well behaved function with **a range from -1 to 1** ($-1 < \sigma(x) < 1$).


$$
\sigma(x) = tanh(x) = \frac{e^x - e^{-x}}{e^x+e^{-x}}
$$
$tanh$ function is just the ratio of some exponential terms. $tanh$ belongs to a family of similar functions or with this characteristic **S shape** called **sigmoids**. Hence, why we use $\sigma$ term for this term.

then our function as: $a^{(1)} = \sigma(w_0a_{0}^{(0)} + w_1a_{1}^{(0)} + b)$

- we have 4 neurons (3 inputs and 1 output) and 3 networks (with 3 weights)

    

then our function as: $a^{(1)} = \sigma(w_0a_{0}^{(0)} + w_1a_{1}^{(0)} + w_2a_{2}^{(0)} + b)$

In general, our expression to take **n inputs** use the summation notation as:
$$
a^{(1)} = \sigma\left(\left(\sum_{j=0}^{n}w_ja_{j}^{(0)}\right) +b\right)
$$
or even we can use beta, that each input has a weight, so we can make a vector of weights and a vector of inputs and then take the dot product to achieve the same effect as above:
$$
a^{(1)} = \sigma(\mathbf{w}\cdot\mathbf{a}^{(0)} + b)
$$

However, we just consider that we have as many inputs with an outputs. Now, we consider the same logic to the outputs (many inputs with many outputs). In this case, we now have twice the number of connectors, each one with its own weighting and each neuron has its own bias.



So our pair of equations to describe this scenario with one for each of the outputs:
$$
\begin{align*}
a_{0}^{(1)} &= \sigma(\mathbf{w}_{0}\cdot\mathbf{a}^{(0)} + b_0) \\
a_{1}^{(1)} &= \sigma(\mathbf{w}_{1}\cdot\mathbf{a}^{(0)} + b_1)
\end{align*}
$$
$$
\iff \boxed{
    a^{(1)} = \sigma(\mathbf{W}^{(1)}\cdot\mathbf{a}^{(0)} + \mathbf{b}^{(1)})
} \quad \text{(compact form)}
$$
in this case, we now hold our two weight vectors in a weight matrix, and our two biases in a bias vector.

**In conclusion**, we consider what our compact equation contains in all its glory. For, what we call **a single-layer neural network** with *m* outputs and *n* inputs. We can fully describe the function it represents with the below equation:


$$

    a^{(1)} = \sigma(\mathbf{W}^{(1)}\cdot\mathbf{a}^{(0)} + \mathbf{b}^{(1)})

$$

However, the neural network would have **one or several** layers of neurons between the inputs and the outputs. We refer to these as **hidden layers**, and they beghave in exactly the same way as we are seen so far.


$$
\begin{align*}
a^{(1)} &= \sigma(\mathbf{W}^{(1)}\cdot\mathbf{a}^{(0)} + \mathbf{b}^{(1)}) \\
a^{(2)} &= \sigma(\mathbf{W}^{(2)}\cdot\mathbf{a}^{(1)} + \mathbf{b}^{(2)})
\end{align*}
$$
With Linear Algebra, we can calculate the outputs of a simple feed forward network:
$$
\boxed{
    a^{(L)} = \sigma(\mathbf{W}^{(L)}\cdot\mathbf{a}^{(L-1)} + \mathbf{b}^{(L)})
}
$$

```python
# Practicing Simple Artificial Neural Networks
import numpy as np
# First set up the network.
sigma = np.tanh
W = np.array([[-2, 4, -1],[6, 0, -3]])
b = np.array([0.1, -2.5])

# Define our input vector
x = np.array([0.3, 0.4, 0.1])
def solve_a1(W1, a0, b1):
    return sigma(W1 @ a0 + b1)
# Calculate the values by hand, and replace a1_0 and a1_1 here (to 2 decimal places)
# (Or if you feel adventurous, find the values with code!)
a1_cal = solve_a1(W, x, b)
a1_0 = a1_cal[0]
a1_1 = a1_cal[1]
a1 = np.array([a1_0, a1_1])
print(a1)
```

When we say that we are training a network, we typically mean using some kind of labelled data, which are pairs of matching inputs and outputs.

**Example:** we build a network to recognise pictures of **faces** and predict if they were **happy**. Whether this image contains a face and whether it was a happy face.

$\implies$ The classic training method is called **back propagation** because it looks first at the output neurons and then it works back through the network

Initially, we will set all of our weights ($\mathbf{w}$) and biases ($\mathbf{b}$) to be a random number. When we pass some data into our network, what we get out will be **meaningless**. However,we can then define **a cost function**, which is simply then sum of the squares of the differences between **the desired output** ($y$), and the output that our **untrained network** currently gives us.
$$
C = \sum_{i}(a_{i}^{(L)} - y_i)^2
$$
If we were somehow able to work out the gradient of $C$ with respect to the variable $w$, at some initial point ($w_0$), then we can simply head in the opposite direction. As below, at the point in the graph, the gradient is positive and therefore increasing $w$ would also increase the cost. So, we should make $w$ smaller to improve our network.



However, in some case our cost function may look some thing more like the wiggly curve, which has several local minima and is more complicated to navigate. Furthermore, this part is just considering one of our weights in isolation. But what we would really like to do is find the minimum of the multi-dimensional hypersurface much like the 2D examples



If we want to head down hill, we will need to build **the Jacobian** by gathering together **the partial derivatives of the cost function** with respect to all of the relevant variables. Now, we just consider again at our simple tưo-node network. At this point, we can immediately write down a chain rule expression for the partial derivative of the cost with respect to either our weight or our bias.


$$
\begin{align*}
a^{(1)} &= \sigma(wa^{(0)} + b)\\
C &= (a^{(1)} - y)^2
\end{align*}
$$
$$
\implies \begin{cases}
\frac{\partial C}{\partial w} = \frac{\partial C}{\partial \mathbf{a}^{(1)}}\frac{\partial \mathbf{a}^{(1)}}{\partial w} \\[10pt]
\frac{\partial C}{\partial b} = \frac{\partial C}{\partial \mathbf{a}^{(1)}}\frac{\partial \mathbf{a}^{(1)}}{\partial b}
\end{cases}
$$
where the $\mathbf{a}^{(1)}$ term which links these two derivatives. However, it is often convenient to make use of an additional term which we call $z^{(1)}$, that will hold our weighted activation plus bias terms.
$$ 
\begin{align*}
z^{(1)} &= wa^{(0)} + b \\[10pt]
\implies a^{(1)} &= \sigma(wa^{(0)} + b) \\
\iff a^{(1)} &= \sigma(z^{(1)})
\end{align*}
$$
This will allow us to think about differentiating the particular sigmoid function that we happened to choose seperately. Therefore, we must include an additional link in our derivative chain.
$$
\implies \begin{cases}
\frac{\partial C}{\partial w} = \frac{\partial C}{\partial \mathbf{a}^{(1)}}\frac{\partial \mathbf{a}^{(1)}}{\partial \mathbf{z}^{(1)}}\frac{\partial \mathbf{z}^{(1)}}{\partial w}\\[10pt]
\frac{\partial C}{\partial b} = \frac{\partial C}{\partial \mathbf{a}^{(1)}}\frac{\partial \mathbf{a}^{(1)}}{\partial \mathbf{z}^{(1)}}\frac{\partial \mathbf{z}^{(1)}}{\partial b}
\end{cases}
$$
We now have the two chains rule expressions we would require to navigate the two dimensional $\mathbf{w} \circ \mathbf{b}$ space in order to minimise the costs of this simple network for a set of training examples. Things are going to get a little bit more complicated when we add more neurons. Fundamentally, we are still just applying the chain rule to link each of our weights and biases back to its effect on the cost, ultimately allowing us to train our network.
$$
\begin{align*}
\mathbf{z}^{(L)} &= \mathbf{W}^{(L)} \cdot \mathbf{a}^{(L-1)} + \mathbf{b}^{(L)} \\
\mathbf{a}^{(L)} &= \sigma(\mathbf{z}^{(L)}) \\
C &= \sum_{i}(\mathbf{a}_{i}^{(L)} - y_{i})^2
\end{align*}
$$

We consider the neural network that has more than 2 layers (which means including at least 1 hidden layers). If we want to calculate the derivative of the cost with respect to the weight or bias of the final layer, then this is the same as previously (but now in vector form):
$$
\frac{\partial C_k}{\partial \mathbf{W}^{(2)}} = \frac{\partial C_k}{\partial \mathbf{a}^{(2)}}\frac{\partial \mathbf{a}^{(2)}}{\partial \mathbf{z}^{(2)}}\frac{\partial \mathbf{z}^{(2)}}{\partial \mathbf{W}^{(2)}}
$$
Moreover, we have $\mathbf{a}^{(2)}$ is the function with respect to $\mathbf{z}^{(2)}$, and $\mathbf{z}^{(2)}$ is the function with respect to $\mathbf{a}^{(1)}$ as follow:
$$
\begin{align*}
\mathbf{a}^{(2)} &= \sigma(\mathbf{z}^{(2)}) \\
\mathbf{z}^{(2)} &= \mathbf{w}^{(2)}\mathbf{a}^{(1)} + \mathbf{b}^{(2)}
\end{align*}
$$
Following th chain rule in multivariate:
$$
\begin{align*}
\frac{\partial\mathbf{a}^{(2)}}{\partial\mathbf{a}^{(1)}} &= \frac{\partial \mathbf{a}^{(2)}}{\partial \mathbf{z}^{(2)}}\frac{\partial \mathbf{z}^{(2)}}{\partial\mathbf{a}^{(1)}} \\[10pt]
\iff \frac{\partial\mathbf{a}^{(2)}}{\partial\mathbf{z}^{(2)}} &= \frac{\partial\mathbf{a}^{(2)}}{\partial\mathbf{a}^{(1)}}\frac{\partial\mathbf{a}^{(1)}}{\partial\mathbf{z}^{(2)}}
\end{align*}
$$
We can calculate the derivative of the function cost with respects to weights of the previous layer:
$$
\frac{\partial C_k}{\partial\mathbf{w}^{(1)}} = \frac{\partial C_k}{\partial\mathbf{a}^{(1)}}\frac{\partial\mathbf{a}^{(1)}}{\partial\mathbf{z}^{(1)}}\frac{\partial\mathbf{z}^{(1)}}{\partial\mathbf{w}^{(1)}}
$$
Or we can use the expression (**the multivariate chain rule**):
$$
\iff \frac{\partial C_k}{\partial\mathbf{w}^{(1)}} = \frac{\partial C_k}{\partial\mathbf{a}^{(2)}}\frac{\partial\mathbf{a}^{(2)}}{\partial\mathbf{a}^{(1)}}\frac{\partial\mathbf{a}^{(1)}}{\partial\mathbf{z}^{(1)}}\frac{\partial\mathbf{z}^{(1)}}{\partial\mathbf{w}^{(1)}}
$$
where $\frac{\partial\mathbf{a}^{(2)}}{\partial\mathbf{a}^{1}}$ itself can be expanded as discussed:
$$
\frac{\partial\mathbf{a}^{(2)}}{\partial\mathbf{z}^{(2)}} = \frac{\partial\mathbf{a}^{(2)}}{\partial\mathbf{a}^{(1)}}\frac{\partial\mathbf{a}^{(1)}}{\partial\mathbf{z}^{(2)}}
$$
This can be generalised to any layer,
$$
\boxed{
    \frac{\partial C_k}{\partial\mathbf{W}^{(i)}} = \frac{\partial C_k}{\partial\mathbf{a}^{(N)}}\underbrace{\frac{\partial\mathbf{a}^{(N)}}{\partial\mathbf{a}^{(N-1)}}\frac{\partial\mathbf{a}^{(N-1)}}{\partial\mathbf{a}^{(N-2)}}\cdots\frac{\partial\mathbf{a}^{(i+1)}}{\partial\mathbf{a}^{(i)}}}_{\text{from layer N to layer i}}\frac{\partial\mathbf{a}^{(i)}}{\partial\mathbf{z}^{(i)}}\frac{\partial\mathbf{z}^{(i)}}{\partial\mathbf{W}^{(i)}}
}
