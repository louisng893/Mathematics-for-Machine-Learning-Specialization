#### Definition of a derivative

gradient (or slope) is "rise over run" $ = \frac{rise}{run} = \frac{y_1 - y_0}{x_1 - x_0}$

$$
\text{gradient at x} = \lim_{\Delta x \to 0}\frac{f(x+\Delta x) - f(x)}{\Delta x}
$$
or in short:
$$ 
\boxed{
    \frac{df}{dx} = f'(x) = \lim_{\Delta x \to 0}\frac{f(x+\Delta x) - f(x)}{\Delta x}
}
$$

*Example:* $f(x) = 3x + 2$
$$
\begin{align*}
f'(x) &= \lim_{\Delta x \to 0}\frac{3(x+\Delta x)+2 -(3x+2)}{\Delta x} \\
&= \lim_{\Delta x \to 0}\frac{3x+3\Delta x+2 -3x+2}{\Delta x} = \lim_{\Delta x \to 0}\frac{3\Delta x}{\Delta x} \\
&= \lim_{\Delta x \to 0}3 = 3
\end{align*}
$$

#### Defferentiation example & cases
- $f(x) = \frac{1}{x}$
$$
\begin{align*}
f'(x) &= \lim_{\Delta x \to 0}\frac{\frac{1}{x + \Delta x} - \frac{1}{x}}{\Delta x} \\
&= \lim_{\Delta x \to 0}\frac{\frac{x - x - \Delta x}{x(x + \Delta x)}} {\Delta x} \\
& = \lim_{\Delta x \to 0}\frac{-1}{x^2 + x\Delta x}\\
& = \frac{-1}{x^2}\\
\end{align*}
$$
- $f(x) = f'(x)$ at $f(x) = f'(x) = f''(x) = f^{(3)}(x) e^x$

### Time saving rules
- Sum rule:
$$
\frac{d}{dx}(f(x) + g(x)) = \frac{df(x)}{dx} + \frac{dg(x)}{dx}
$$
- Power rule: If $f(x) = ax^b$ then $f(x) = abx^{b-1}$
- Product rule:
$$
\begin{align*}
\frac{d}{dx}(f(x)g(x)) &= \frac{df(x)}{dx}g(x) + f(x)\frac{dg(x)}{dx} \\
\iff [f(x)g(x)]' &= f'(x)g(x) + f(x)g'(x)
\end{align*}
$$
- Chain rule: If $f = f(g)$ and $g = g(x)$, then
$$
\frac{df}{dx} = \frac{df}{dg}\frac{dg}{dx}
$$

*Example:* $f(x) = \frac{sin(2x^5+3x)}{e^{7x}} = (sin(2x^5 + 3x))e^{-7x}$
$$
\begin{align*}
\frac{df}{dx}  &= \frac{dg}{dx}h + g\frac{dh}{dx} \quad\text{(product rule)}\\
&= (10x^4 + 3)cos(2x^5+3x)e^{-7x} + sin(2x^5 + 3x)(-7e^{-7x}) \quad\text{(chain rule)} \\
&= e^{-7x}[(10x^4 +3)cos(2x^5 + 3x) - 7sin(2x^5 + 3x)] \\[10pt]
&= \frac{(10x^4 + 3)cos(2x^5 + 3x)}{e^{7x}} - 7f(x)
\end{align*}

### Jacobians - vectors of derivatives

#### The Jacobian
We consider a single function with many variables $f(x_1, x_2, x_3,...)$, then the Jacobian is simply a vector where each entry is the partial derivative of $f$ (with row vectors rather than column vectors).
$$
J = \begin{bmatrix}\frac{\partial f}{\partial x_1}&\frac{\partial f}{\partial x_2}&\frac{\partial f}{\partial x_3}&\cdots\end{bmatrix}
$$
*Example:* $f(x, y, z) = x^2y + 3z$
$$
\left. 
\begin{align*}
\frac{\partial f}{\partial x} &= 2xy \\
\frac{\partial f}{\partial y} &= x^2 \\
\frac{\partial f}{\partial z} &= 3
\end{align*}
\right\} \qquad J = \begin{bmatrix}2xy&x^2&3\end{bmatrix}
$$
For a specific given $x$, $y$, and $z$ will return a vector pointing in the direction of steepest slope of this function. For example, at $J(0, 0, 0) = \begin{bmatrix}0&0&3\end{bmatrix}$. 
> The larger the magnitude of the Jacobian vectors, the steeper the slope.

#### Jacobian Applied
$$
\boxed{
    J = \begin{bmatrix} \frac{\partial u}{\partial x}&\frac{\partial u}{\partial y}\\\frac{\partial v}{\partial x}&\frac{\partial v}{\partial y}\end{bmatrix}
    }
$$
For the pair of equations: 
- $u(x, y) = x - 2y$
- $v(x, y) = 3y - 2x$

we can calculate the Jacobian matrix as: $J = \begin{bmatrix}1&-2\\-2&3\end{bmatrix}$, there are no longer any variables in the Jacobian matrix, it means both $u$ and $v$ are linear functions of $x$ and $y$, so the gradient must be constant everywhere. Also, this matrix is just the linear transformation from $x\circ y$ space to $u\circ v$ space.

If we consider a vector $\begin{pmatrix}2\\3\end{pmatrix}$, we get the following $\begin{bmatrix}1&-2\\-2&3\end{bmatrix}\begin{pmatrix}2\\3\end{pmatrix} = \begin{pmatrix}-4\\5\end{pmatrix}$.

Sometime functions will be highly nonlinear, much more complicated. However, they may still be smooth, which means that if we zoom in close enough, we can consider each little region of space to be approximately linear. Therefore, by adding up all the contributions from the Jacobian determinants at each point in space, we can still calculate the change in the size of a region after transformation.

***Example:*** transform between Cartesian and Polar coodinate systems



If we have a vector expressed in terms of a radius $r$, and the angle up $\theta$ from the x-axis. But we would like them expressed in terms of $x$ and $y$ instead.
- $x(r, \theta) = rcos(\theta)$
- $y(r, \theta) = rsin(\theta)$

Then the Jacobian matrix is
$$ 
\begin{align*}
J &= \begin{bmatrix}\frac{\partial x}{\partial r}&\frac{\partial x}{\partial\theta}\\\frac{\partial y}{\partial r}&\frac{\partial y}{\partial\theta}\end{bmatrix}\\
& =\begin{bmatrix}cos(\theta)&-rsin(\theta)\\ sin(\theta)&rcos(\theta)\end{bmatrix} \\[10pt]
\implies |J| &= r(cos^2(\theta) + sin^2(\theta)) \\
&= r
\end{align*}
$$


### The sandpit game

#### The sandpit game
We consider a function $f(x, y) = e^{-(x^2+y^2)}$, then we can easily find the Jacobian as discussed $J = \begin{bmatrix}-2xe^{-(x^2 + y^2)}&-2ye^{-(x^2 + y^2)}\end{bmatrix} = 0$ (it also means $f'(x, y) = 0$), solving this equation we get $x = 0$ and $y = 0$.

Therefore, with this function or more other complicated functions, we have, at the critical points - including minima, maxima, or saddle point,the Jacobian (gradient) equals to zero ($J=0$). The direction of the vector points to the direction of the steepest ascent (uphill), and its length indicates how steep the slope is (at a point). In general, if we stand near **the local minima** (the trough), the gradient vectors around us will **point away from** us. Conversely, near **the local maxima** (the peak), around gradient vectors will **point toward** us.

#### The Hessian
For the Jacobian, we collected together all *the first order* derivatives of a function into a vector. We are going to collect all of *the second order* derivatives together into a matrix, which for a function of n variables - **The Hessian**.

$$
\boxed{
    \mathbf{H} = \begin{bmatrix}\frac{\partial^2 f}{\partial x_{1}^{2}}&\frac{\partial^2 f}{\partial x_{1}x_{2}}&\cdots&\frac{\partial^2 f}{\partial x_{1}x_{n}}\\[5pt]\frac{\partial^2 f}{\partial x_{2}x_{1}}&\frac{\partial^2 f}{\partial x_{2}^{2}}&\cdots&\frac{\partial^2 f}{\partial x_{2}x_{n}}\\[5pt]\vdots&\vdots&\ddots&\vdots\\[5pt]\frac{\partial^2 f}{\partial x_{n}x_{1}}&\frac{\partial^2 f}{\partial x_{n}^{2}}&\cdots&\frac{\partial^2 f}{\partial x_{n}^{2}}\end{bmatrix}
}
$$
We can write it using an abbreviated notation style
$$
\iff 
\mathbf{H} = \begin{bmatrix}\partial_{x_1x_1}f&\partial_{x_1x_2}f&\cdots&\partial_{x_1x_n}f\\\partial_{x_2x_1}f&\partial_{x_2x_2}f&\cdots&\partial_{x_2x_n}f\\\vdots&\vdots&\ddots&\vdots\\\partial_{x_nx_1}f&\partial_{x_nx_2}f&\cdots&\partial_{x_nx_n}f\end{bmatrix}
$$
The Hessian matrix is an n by n square matrix, where n is the number of variables in our function $f$.

*Example:* $f(x, y, z) = x^2yz$ (find the Jacobian first, then the Hessian)
$$
J =\begin{bmatrix}2xyz&x^2z&x^2y\end{bmatrix} \\[10pt]
H = \begin{bmatrix}2yz&2xz&2xy\\2xz&0&x^2\\2xy&x^2&0\end{bmatrix}
$$
Our Hessian matrix is symmetrical across the leading diagonal. This will always be true if the function is continuous, meaning that it has no sudden step changes.

**For example:** We consider two cases:
- Case 1: $f(x, y) = x^2 + y^2$, then the Jacobian is $J = \begin{bmatrix}2x&2y\end{bmatrix}$, and the Hessian is $\mathbf{H} = \begin{bmatrix}2&0\\0&2\end{bmatrix}$. Therefore, the determinant of the Hessian ($|\mathbf{H}| = 4$) is **positive**, we are dealing with **either a maximum or a minimum** (or an optimum point). Moreover, the first term at top left of the Hessian, ($\mathbf{H} = \begin{bmatrix}\mathbf{2}&0\\0&2\end{bmatrix}$), is also **positive** then we have got **a minimum** (when it is negative, we have got a maximum).

    

- Case 2: $f(x, y) = x^2 - y^2$, then the Jacobian is $J = \begin{bmatrix}2x&-2y\end{bmatrix}$, and the Hessian is $\mathbf{H} = \begin{bmatrix}2&0\\0&-2\end{bmatrix}$. Therefore, the determinant of the Hessian ($|\mathbf{H}| = -4$) is **negative**, we are **not** dealing with **a maximum or a minimum**. At this point, the gradient is 0, but with slopes coming down towards it in one direction, but up towards it in the other - **the saddle point**.

## Chapter 5: Intro to optimisation
